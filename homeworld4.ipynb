bài 18:Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b

```python
a, b = map(int, input().split())
so_min = min(a, b)

for i in range(1, so_min + 1):
    if a % i == 0 and b % i == 0:
        print(i, end = " ")
```

bài 19: Viết chương trình Python sử dụng một trong 2 loại vòng lặp for hoặc while để in ra các số dương chẵn nhỏ hơn hoặc bằng n (n <= 1000) 

```python
n = int(input())

for i in range(2, n + 1, 2):
    print(i, end = " ")

print()

j = 2
while j <= n:
    print(j, end = " ")
    j += 2
```

bài 20:Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không

```python
n = int(input())

while n % 2 == 0:
    n //= 2

if n == 1:
    print("yes")
else:
    print("no")
```

bài 21: Nhập vào số nguyên dương n, tính tổng các chữ số của n

```python
n = int(input())
tong = 0

while n > 0:
    tong += n % 10
    n //= 10

print(tong)
```

bài 22: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ

```python
n = int(input())
c = 0
l = 0

while n > 0:
    i = n % 10
    if i % 2 == 0:
        c += 1
    else:
        l += 1
    n //= 10

print(f"chãn: {c}, lẻ: {l}")
```

Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n

```python
n = int(input())
s = 0
k = 0

while s + (k + 1) < n:
    k += 1
    s += k

print(k)
```

bài 24:Nhập vào A, tìm n nhỏ nhất sao cho 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 


```python
a = int(input())
s = 0
n = 0

while s < a:
    n += 1
    s += 1/n

print(n)
```

bài 25: Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập 


```python
n = int(input())
n_max = None
n_min = None

while True:
    if n_max == None or n_max < n:
        n_max = n
    if n_min == None or n_min > n:
        n_min = n

    n = int(input())
    
    if n == -1:
        break

print(n_max, n_min)
```

bài 26: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A

```python
a = int(input())
x = 1
y = 1

while x + y < a:
    temp = x + y
    x = y
    y = temp

print(y)
```

bài 27: Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ (quy định là chuỗi không có ký tự đặc biệt, không số, không có dấu câu, chỉ có ký tự chữ và khoảng trắng) Dùng cách: for trên index; KHÔNG dùng hàm split(), len()

```python
c = input()
kq = 0
# length = 0

# for _ in c:
#     length += 1

for i in range(len(c)):
    if c[i] != " " and (i == 0 or c[i-1] == " "):
        kq += 1

print(kq)
```

bài 28: Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi. Dùng cách: for trên index; KHÔNG dùng hàm split() 

```python
c = input()
kq = ""

for i in range(len(c)):
    if c[i] == " ":
        break
    kq += c[i]

print(kq)
```

bài 30:Nhập vào một chuỗi có dạng 3 số nguyên, mỗi số nguyên cách nhau một dấu phẩy, hãy tính tổng 3 số nguyên đó
VD:
Nhập: 3, 12, 15
Tổng: 30 


```python
n = input()
tong = 0

for i in n.split(","):
    tong += int(i)

print(tong)
```

bài 31:Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in thường, bao nhiêu ký tự số 

```python
c = input()
hoa = 0
thuong = 0
so = 0

for i in c:
    if i.isupper():
        hoa += 1
    elif i.islower():
        thuong += 1
    elif i.isnumeric():
        so += 1

print(f"thường: {thuong}, hoa: {hoa}, so: {so}")
```
