bài 6:Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn 
[a, b]. Mỗi dòng là một bộ test (a b).

```python
a, b = map(int, input().split())
tong = 0
for i in range(a, b+1):
    if i < 2:
        continue

    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            break
    else:
        tong += i
            
print(tong)
```

bài 8: Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.

```python
def reverse(n):
    rev_n = 0
    while n > 0:
        rev_n = rev_n*10 + (n % 10)
        n //= 10
    return rev_n

def palindrome(n):
    return reverse(n) == n

n = int(input())
dem = 0

while not palindrome(n):
    n = n + reverse(n)
    dem += 1

print(dem, reverse(n))
```

bài 9: Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu cách). Nếu không có, in ra (no number).

```python
def ktra(n):
    while n > 0:
        so_cuoi = n % 10
        con_lai = n // 10
        
        while con_lai > 0:
            so_khac = con_lai % 10
            if so_cuoi == so_khac:
                return False
            con_lai = con_lai // 10
        
        n = n // 10
    return True

n = int(input())
co_so = False
da_in = False

i = 1
while i * i <= n:
    so_cp = i * i
    if ktra(so_cp):
        if da_in:
            print(" ", end="")
        print(so_cp, end="")
        co_so = True
        da_in = True
    i = i + 1

if not co_so:
    print("no number")
```

bài 10:Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất

```python
n = int(input())
cal_max = 0
best_x = 0

for x in range(1, n + 1):
    temp = x
    curr_len = 1

    while temp != 1:
        if temp % 2 == 0:
            temp //= 2
        else:
            temp = 3 * temp + 1
        curr_len += 1

    if curr_len > cal_max:
        cal_max = curr_len
        best_x = x

print(best_x, cal_max)
```

bài 13:
Hai số nguyên dương tạo thành một cặp số thân thiết khi chúng tuân theo quy luật: 
Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại.
Ví dụ 220 và 284 là một cặp số thân thiết 220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10,11, 20, 44, 55, 110. Tổng của 11 ước này cho chúng ta 284 ,284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng vừa đúng bằng 220 
Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không. Nếu có, in ra màn hình "true", ngược lại, in ra "false"

```python
a, b = map(int, input().split())
tong_uoc_a = 0
tong_uoc_b = 0

for i in range(1, a//2 + 1):
    if a % i == 0:
        tong_uoc_a += i

for j in range(1, b//2 + 1):
    if b % j == 0:
        tong_uoc_b += j

if tong_uoc_a == b and tong_uoc_b == a:
    print("True")
else:
    print("False")
```

bài 14:Cho hai số m và n là hai số nguyên dương nhập từ bàn phím. Viết chương trình tìm ước chung lớn nhất của hai số m và n.

```python
m, n = map(int, input().split())
so_min = min(m, n)
ucln = 0

for i in range(1, so_min + 1):
    if m % i == 0 and n % i == 0:
        if ucln < i:
            ucln = i

print(ucln)

```
