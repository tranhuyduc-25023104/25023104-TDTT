bài 1: Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]


```python
l = map(int, input().split())
ls = []
for i in l:
    if i not in ls:
        ls.append(i)
print(ls)
```

bài 2: Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị trí i), giữ nguyên thứ tự

```python
l = map(int,input().split())
tong = 0
a = []
for i in l:
    tong += i
    a.append(tong)
print(a)
```

bài 4: Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó


```python
dic = input().split()
dickq = {}
for i in dic:
    key, value = i.split(":")
    if key in dickq:
        dickq[key].append(value)
    else:
        dickq[key] = [value]

print(dickq)
```

bài 5:Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros': ...} (đếm số >0, <0, =0)

```python
l = list(map(int, input().split()))
dic = {'positives': 0, 'negatives': 0, 'zeros': 0}

for i in l:
    if i > 0:
        dic['positives'] += 1
    elif i < 0:
        dic['negatives'] += 1
    else:
        dic['zeros'] += 1

print(dic)
```

bài 6:Viết chương trình nhập vào một danh sách các số nguyên, sau đó in ra tổng các số trong list đó
