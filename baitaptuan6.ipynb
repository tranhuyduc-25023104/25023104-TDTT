bài 1: Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]


```python
l = map(int, input().split())
ls = []
for i in l:
    if i not in ls:
        ls.append(i)
print(ls)
```
