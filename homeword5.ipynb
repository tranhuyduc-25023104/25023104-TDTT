bài 16:Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên 

```python
def amstronhg(n):
    n_tam = n
    m = len(str(n))
    tong = 0
    while n_tam > 0:
        tong += (n_tam % 10) ** m
        n_tam //= 10
    return tong == n

n = int(input())
print(amstronhg(n))
```

bài 18: VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.

```python
def doi_xung(n):
    temp = n
    so_dao_nguoc = 0

    while temp > 0:
        so_dao_nguoc = (so_dao_nguoc * 10) + (temp % 10)
        temp //= 10
    
    return n == so_dao_nguoc

n = int(input())
print(doi_xung(n))
```

bài 19:VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên

```python
def so_ky_dieu(n):
    while n > 9:
        tong_chu_so = 0
        while n > 0:
            tong_chu_so += n % 10
            n //= 10
        n = tong_chu_so 
        
    return n == 1

n = int(input())
print(so_ky_dieu(n))
```

bài 20:Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def detect_number_type(s):
    if s.isdigit():
        return "int"
    if s.count(".") == 1:
        if s.replace(".", "", 1).isdigit():
            return "float"
    return "neither"

s = input()
print(detect_number_type(s))
```

bài 23:Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

```python
def quadratic_roots_count(a, b, c):
    delta = b*b - 4*a*c
    if delta > 0:
        return (-b + delta**0.5) / (2 * a), (-b - delta**0.5) / (2 * a)
    elif delta == 0:
        return -b / (2 * a)
    else:
        return "pt vô nghiệm"

a, b, c = map(int, input().split())
print(quadratic_roots_count(a, b, c))
```

bài 24:Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.

```python
def are_collinear(x1,y1,x2,y2,x3,y3):
    ABx , ABy = x2 - x1, y2 - y1
    ACx , ACy = x3 - x1, y3 - y1

    return (ABx * ACy) - (ABy * ACx) == 0

x1, y1 = map(float, input().split())
x2, y2 = map(float, input().split())
x3, y3 = map(float, input().split())
print(are_collinear(x1,y1,x2,y2,x3,y3))
```

bài 25:Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. 
VD: quad = [(0,0), (4,0), (4,3),(0,3)], p = (2,1)

```python
def point_in_quadrilateral(quad, p):
    xp, yp = p
    
    x1, y1 = quad[0]
    x2, y2 = quad[1]
    x3, y3 = quad[2]
    x4, y4 = quad[3]
    
    cross1 = (x2 - x1) * (yp - y1) - (y2 - y1) * (xp - x1)  
    cross2 = (x3 - x2) * (yp - y2) - (y3 - y2) * (xp - x2)
    cross3 = (x4 - x3) * (yp - y3) - (y4 - y3) * (xp - x3)
    cross4 = (x1 - x4) * (yp - y4) - (y1 - y4) * (xp - x4)
    
    tat_ca_duong = (cross1 >= 0 and cross2 >= 0 and cross3 >= 0 and cross4 >= 0)
    tat_ca_am    = (cross1 <= 0 and cross2 <= 0 and cross3 <= 0 and cross4 <= 0)
    
    return tat_ca_duong or tat_ca_am

A = tuple(map(float, input().split()))
B = tuple(map(float, input().split()))
C = tuple(map(float, input().split()))
D = tuple(map(float, input().split()))

quad = [A, B, C, D]
p = tuple(map(float, input().split()))
print(point_in_quadrilateral(quad, p)) 
```

bài 28:Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3)

```python
import math

def tinh_goc_ABC(A, B, C):
    xa, ya = A
    xb, yb = B
    xc, yc = C

    xba, yba = xa - xb, ya - yb
    xbc, ybc = xc - xb, yc - yb

    tich_vo_huong = xba * xbc + yba * ybc
    BA = (xba**2 + yba**2)**0.5
    BC = (xbc**2 + ybc**2)**0.5

    goc_radian = math.acos(tich_vo_huong/(BA * BC))
    goc_do = math.degrees(goc_radian)

    return goc_do

A = tuple(map(float, input().split()))
B = tuple(map(float, input().split()))
C = tuple(map(float, input().split()))
print(tinh_goc_ABC(A, B, C))
```

bài 30: Viết hàm tính hệ số nhị thức C(n, k).

```python
def tinh_he_so_nhi_thuc(n, k):
    if k > n // 2:
        k = n - k

    tu = 1
    mau = 1

    for i in range(1, k + 1):
        tu *= (n - i + 1)
        mau *= i

    return tu // mau

n, k = map(int, input().split())
print(tinh_he_so_nhi_thuc(n ,k))
```

bài 34:Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n):
    if n <= 0:
        return 0
    
    tong = 0
    for x in range(1, n + 1):
        s = str(x)
        if len(s) == len(set(s)):
            tong += x
            
    return tong

n = int(input())
print(sum_1_to_n(n))
```
