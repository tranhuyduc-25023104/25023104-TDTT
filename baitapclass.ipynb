bài 1:
[Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diệntích toàn phần) và thể tích của hình trụ
Đầu vào
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách.
Đầu ra
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử dụng lại những hàm vừa viết, π=3.14).
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy)

```python
class Cylinder:
    def __init__(self, rr, hh):
        self.r = rr
        self.h = hh
    def dien_tich(self):
        return (2 * (self.r ** 2) * 3.14) + (2 * 3.14 * self.r * self.h)
    def the_tich(self):
        return (3.14 * (self.r ** 2) * self.h)
    
dauvao = list(map(float, input().split()))
rr = dauvao[0]
hh = dauvao[1]
cylinder1 = Cylinder(rr, hh)
rs1 = cylinder1.dien_tich()
rs2 = cylinder1.the_tich()
print(f"{rs1:.2f} {rs2:.2f}")
```

ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.
Đầu vào
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng
dd/mm/yyyy
.
Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class Date:
  def __init__(self, dd, mm ,yy):
    self.d = dd
    self.m = mm
    self.y = yy
  def check_valid(self):
    max_day = -1

    match self.m:
      case 1| 3| 5| 7| 8| 10| 12:
        max_day = 31
      case 4| 6| 9| 11:
        max_day = 30
      case 2:
        if (self.y % 4 == 0 and self.y % 100 != 0) and self.y % 400 == 0:
          max_day = 29
        else:
          max_day = 28
    return (self.y > 0) and (1 <= self.d <= max_day)

  def get_nextday(self):
    if self.check_valid():
        nextday = Date(self.d + 1, self.m, self.y)
    if nextday.check_valid():
        return nextday
    nextday = Date(1, self.m + 1, self.y)
    if nextday.check_valid():
        return nextday
    return Date(1, 1, self.y+1)

  def __str__(self):
    return f"{self.d:02d}/{self.m:02d}/{self.y}"

day1 = Date(29, 1, 2000)
day2 = Date(29, 2, 2001)
day3 = Date(21, 4, 2001)
print(day1.check_valid())
print(day2.check_valid())
print(day3.check_valid())


print(day3.get_nextday)
      
```

Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính, Mỗi phương thức thực hiện một phép tính và trả về kết quả tương ứng

```python
class calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2
    
    def set_numbers(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2
    
    def subtract(self):
        return self.num1 - self.num2
    
    def product(self):
        return self.num1 * self.num2
    
    def divide(self):
        if self.num2 != 0:
            return self.num1 / self.num2
        else:
            return "lỗi"
    
    def power(self):
        return self.num1 ** self.num2
    
    def mod(self):
        if self.num2 != 0:
            return self.num1 % self.num2
        else:
            return "lỗi"
        
n1 = float(input())
n2 = float(input())
calc = calculator(n1, n2)

while True:
    print("\n--- MENU MÁY TÍNH ---")
    print("1. Cộng (add)")
    print("2. Trừ (subtract)")
    print("3. Nhân (product)")
    print("4. Chia (divide)")
    print("5. Hàm mũ (power)")
    print("6. Lấy phần dư (mod)")
    print("7. Cập nhật số mới (set_numbers)")

    choice = input("chọn phép tính (1 - 7): ")

    if choice == '1':
        print(f"Kết quả: {calc.add()}")
    elif choice == '2':
        print(f"Kết quả: {calc.subtract()}")
    elif choice == '3':
        print(f"Kết quả: {calc.product()}")
    elif choice == '4':
        print(f"Kết quả: {calc.divide()}")
    elif choice == '5':
        print(f"Kết quả: {calc.power()}")
    elif choice == '6':
        print(f"Kết quả: {calc.mod()}")
    elif choice == '7':
        n1 = float(input("Nhập số thứ nhất mới: "))
        n2 = float(input("Nhập số thứ hai mới: "))
        calc.set_numbers(n1, n2)
        print("Đã cập nhật số mới!")
    else:
        print("Lựa chọn không hợp lệ.")

    quit_choice = input("Bạn có muốn thoát? (Gõ 'yes' để thoát): ").strip().lower()
    if quit_choice == "yes":
        print("Tạm biệt!")
        break
```

bài 5:[ShoppingCart]
Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi sản phẩm là một từ điển có name (tên) và price(giá).
Với các phương thức sau:
1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
4. Kiểm tra xem giỏ hàng có rỗng không
5. Tính tổng tiền giỏ hàng
6. Hiển thị toàn bộ sản phẩm trong giỏ
7. Xóa toàn bộ giỏ hàng
Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7.
Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có muốn thoát khỏi chương trình không?"

```python
class shoppingCart:
    def __init__(self):
        self.items = []

    def add_product(self, name, price):
        product = {"name": name, "price": price}
        self.items.append(product)
        print(f"đã thêm: {name} - {price}")
    
    def remove_product(self, name):
        found = False
        for item in self.items:
            if item['name'].lower() == name.lower():
                self.items.remove(item)
                found = True
                print(f"Đã xóa '{name}' khỏi giỏ hàng.")
                break
        if not found:
            print(f"Không tìm thấy sản phẩm '{name}'.")


    def is_empty(self):
        return len(self.items) == 0
    
    def total_price(self):
        tong = 0
        for item in self.items:
            tong += item["price"]
        
        return tong
    
    def display_all(self):
        if self.is_empty():
            print("giỏ hàng trống")
        else:
            print("--- Danh sachsanr phẩm")
            temp = 1
            for item in self.items:
                print(f"{temp}. {item["name"]}: {item["price"]}")

    def clear_cart(self):
        self.items = []
        print("đã xoá toàn bộ giả hàng")

cart = shoppingCart()

while True:
    print("\n--- MENU SHOPPING CART ---")
    print("2. Thêm sản phẩm")
    print("3. Xóa sản phẩm theo tên")
    print("4. Kiểm tra giỏ hàng rỗng")
    print("5. Tính tổng tiền")
    print("6. Hiển thị giỏ hàng")
    print("7. Xóa toàn bộ giỏ hàng")
    
    choice = input("Nhập lựa chọn của bạn (2-7): ")

    if choice == '2':
        name = input("Nhập tên sản phẩm: ")
        price = float(input("Nhập giá sản phẩm: "))
        cart.add_product(name, price)
    elif choice == '3':
        name = input("Nhập tên sản phẩm cần xóa: ")
        cart.remove_product(name)
    elif choice == '4':
        if cart.is_empty():
            print("Giỏ hàng đang rỗng.")
        else:
            print("Giỏ hàng đang có sản phẩm.")
    elif choice == '5':
        print(f"Tổng tiền: {cart.total_price()}")
    elif choice == '6':
        cart.display_all()
    elif choice == '7':
        cart.clear_cart()
    else:
        print("Lựa chọn không hợp lệ.")

    exit_confirm = input("Có muốn thoát khỏi chương trình không? (nhập 'yes' để thoát): ").lower()
    if exit_confirm == 'yes':
        print("Cảm ơn bạn đã sử dụng dịch vụ!")
        break
```

bài 6: Yêu cầu:
Xây dựng lớp Rectangle(w, h) với các phương thức:
area() trả về diện tích.
perimeter() trả về chu vi.
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h), không trả về gì.
Input: w h k (số thực dương)
Output: area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân)

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
    
    def area(self):
        return self.w * self.h
    
    def perimeter(self):
        return 2 * self.w + 2 * self.h
    
    def scale(self, k):
        self.w *= k
        self.h *= k

w, h, k = map(float, input().split())
rect = Rectangle(w, h)
rect.scale(k)
print(f"{rect.area():.2f}")
print(f"{rect.perimeter():.2f}")
```

bài 7:Yêu cầu:
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:
Rút gọn phân số trong __init__ (quy ước mẫu > 0).
Các phương thức add/sub/mul/div trả về phân số đã rút gọn. __str__() in dạng a/b.
Input: a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})
Output: Phân số kết quả đã rút gọn 


```python
class Fraction:
    def __init__(self, num, den):
        if den == 0:
            return
        
        def find_gcd(x, y):
            x, y = abs(x), abs(y)
            while y:
                x, y = y, x % y
            return x

        g = find_gcd(num, den)
        self.num = num // g
        self.den = den // g
        
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den

    def add(self, other):
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    def sub(self, other):
        new_num = self.num * other.den - other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    def mul(self, other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)

    def div(self, other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)

    def __str__(self):
        return f"{self.num}/{self.den}"

a, b = map(int,input().split())
op = input()
c, d = map(int,input().split())

f1 = Fraction(a, b)
f2 = Fraction(c, d)
    
if op == '+':
    result = f1.add(f2)
elif op == '-':
    result = f1.sub(f2)
elif op == '*':
    result = f1.mul(f2)
elif op == '/':
    result = f1.div(f2)
        
print(result)
```

bài 8:Yêu cầu
Lớp BankAccount(owner, balance) với:
deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch.__str__() in owner: balance.
Input:
Dòng 1: owner initial_balance
Dòng 2: n (số lệnh)
Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.
Output:
Số dư cuối cùng (in số thực, 2 chữ số)

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, other):
        self.balance += other

    def withdraw(self, other):
        if other <= self.balance:
            self.balance -= other

    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"
    
owner = input()
initial_balance = float(input())
account = BankAccount(owner, initial_balance)

n = int(input())
for _ in range(n):
        command = input().split()
        action = command[0]
        amount = float(command[1])

        if action == "DEPOSIT":
            account.deposit(amount)
        elif action == "WITHDRAW":
            account.withdraw(amount)

print(f"{account.balance:.2f}")
```

bài 9:Yêu cầu
Lớp Student(name) với:
add_score(subject, score)
avg() trả về điểm trung bình.
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8, "Average" nếu 5–<6.5, ngược lại "Poor".
Input:
Dòng 1: name
Dòng 2: n
n dòng tiếp: subject score (score là số thực 0–10)
Output:
name avg rank (avg 2 chữ số). 

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.score = []

    def add_score(self, subject, score):
        self.subject = subject
        self.score.append(score)
    
    def avg(self):
        return sum(self.score) / len(self.score)
    
    def rank(self):
        if self.avg() >= 8:
            return "Excellent"
        elif self.avg() >= 6.5:
            return "Good"
        elif self.avg() >= 5:
            return "Average"
        else:
            return "Poor"
        
name = input()
student = Student(name)
n = int(input())
for _ in range(n):
    line = input().split()
    subject = " ".join(line[:-1])
    score = float(line[-1])
    student.add_score(subject, score)

print(f"{student.name} {student.avg():.2f} {student.rank()}")
```

# lớp Book(title, author, year).
# Lớp Library() với:
- add(book)
- count_by_author(name)
- find_by_year(y) trả về số sách xuất bản năm y.
# Input:
 Dòng 1: n (số thao tác),
 n dòng:
- ADD title;author;year
- COUNT author (đếm theo tác giả)
- COUNTYEAR year (đếm theo năm)
# Output:
Mỗi lệnh COUNT/COUNTYEAR in một dòng kết quả.

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

class Library:
    def __init__(self):
        self.books = []

    def add(self, book):
        self.books.append(book)

    def count_author(self, name):
        count = 0
        for book in self.books:
            if book.author == name:
                count += 1
        return count
    
    def count_year(self, y):
        count = 0
        for book in self.books:
            if book.year == y:
                count += 1
        return count
    
library = Library()

n = int(input())
for _ in range(n):
    line = input()
    if line[0:4] == "ADD ":
        content = line[4:]

        parts = []
        current_part = ""
        for char in content:
            if char == ";":
                parts.append(current_part)
                current_part = ""
            else:
                current_part += char
        parts.append(current_part)

        if len(parts) == 3:
            title = parts[0]
            author = parts[1]
            year = parts[2]

            new_book = Book(title,author,year)
            library.add(new_book)

    elif line[0:6] == "COUNT ":
        search_author = line[6:]
        result = library.count_author(search_author)
        print(result)

    elif line[0:10] == "COUNTYEAR ":
        search_year = line[10:]
        result = library.count_year(search_year)
        print(result)

    
```

bài 11:
Bạn được cho bốn điểm A, B, C và D trong hệ tọa độ Đề-các 3 chiều.
Nhiệm vụ của bạn là tính và in ra góc giữa hai mặt phẳng tạo bởi các điểm:
- mặt phẳng (A, B, C)
- mặt phẳng (B, C, D)
Góc cần in ra được ký hiệu là PHI, và phải dùng độ (degrees),
không dùng radian.
Dạng Input:
Một dòng input chứa 3 số thực, ngăn cách bằng dấu cách, biểu
diễn:
tọa độ X tọa độ Y tọa độ Z của một điểm.
Dạng Output:
In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập
phân. 

```python
import math

class Points:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    def __sub__(self, other):
        return Points(self.x - other.x, self.y - other.y, self.z - other.z)
    
    def dot(self, other):
        return self.x*other.x + self.y*other.y +self.z*other.z
    
    def cross(self, other):
        return Points(self.y * other.z - self.z * other.y, self.z * other.x - self.x * other.z, self.x * other.y - self.y * other.x)
    
    def absolute(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)
    
points = []

for i in range(4):
    line = input().split()
    points.append(Points(float(line[0]), float(line[1]), float(line[2])))

a = points[0]
b = points[1]
c = points[2]
d = points[3]

ab = b - a
bc = c - b
cd = d - c

x = ab.cross(bc)
y = bc.cross(cd)

dot_val = x.dot(y)
abs_dot = abs(dot_val)

cos_phi = abs_dot / (x.absolute() * y.absolute())
cos_phi = max(-1.0, min(1.0, cos_phi))
phi_radian = math.acos(cos_phi)
phi_degree = math.degrees(phi_radian)

print(f"{phi_degree:.2f}")
```

Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo của một số phức. Các số trong đầu vào có giá trị tuyệt đối không vượt quá \(100\).
Đầu ra:
In ra màn hình hai số biểu diễn phần thực và phần ảo của tích
hai số phức.


```python
class ComplexNumber:
    def __init__(self, realPart, imaginaryPart):
        self.realPart = realPart
        self.imaginaryPart = imaginaryPart
    
    def multiply(self, other):
        new_real = self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart
        new_imaginary = self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart
        return ComplexNumber(new_real, new_imaginary)
    
r1, i1 = map(float, input().split())
r2, i2 = map(float, input().split())
z1 = ComplexNumber(r1, i1)
z2 = ComplexNumber(r2, i2)

result = z1.multiply(z2)

if result.realPart % 1 == 0:
    real_out = int(result.realPart)
else:
    real_out = result.realPart

if result.imaginaryPart % 1 == 0:
    imag_out = int(result.imaginaryPart)
else:
    imag_out = result.imaginaryPart

print(real_out, imag_out)
```

bài 13:[Triangle]
Khai báo cấu trúc dữ liệu tam giác \(Triangle\) chứa \(3\) biến \(egde1,\,edge2,\,edge3\) kiểu \(int\) là độ dài \(3\) cạnh của tam giác đó.
Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm double getArea() trả về diện tích của tam giác đó (nếu tam giác không hợp lệ, trả về \(-1\)). Hoàn thiện chương trình dưới đây nhập vào \(3\) cạnh của một tam giác và in ra diện tích của tam giác đó.
Đầu vào:
Một dòng duy nhất từ bàn phím chứa \(3\) số nguyên là độ dài \(3\) cạnh của tam giác
Đầu ra:
In ra màn hình diện tích tam giác (làm tròn đến\(2\) chữ số thập phân sau dấu phẩy). In ra \(invalid\) nếu \(3\) số nhập vào không phải là \(3\) cạnh của một tam giác 


```python
class Triangle:
    def __init__(self, egde1, egde2, egde3):
        self.egde1 = egde1
        self.egde2 = egde2
        self.egde3 = egde3

    def getArea(self):
        a, b, c = self.egde1, self.egde2, self.egde3
        if (a + b > c) and (a + c > b) and (b + c > a):
            p = (a + b +c) / 2
            s = (p * (p - a) * (p - b) * (p - c)) ** 0.5
            return s
        else:
            return -1
        
egde1 , egde2, egde3 = map(int, input().split())
triangle = Triangle(egde1, egde2, egde3)
s = triangle.getArea()

if s == -1:
    print("invalid")
else:
    print(f"{s:.2f}")
```

bài 14:Yêu cầu:
Xây dựng lớp Animal với phương thức speak().
Xây dựng hai lớp kế thừa Dog và Cat, mỗi lớp ghi đè speak()
để trả về âm thanh tương ứng:
- Dog -> Woof
- Cat -> Meow
Input:
Một dòng chứa tên lớp: Dog hoặc Cat.
Output:
In ra kết quả của speak()

```python
class Animal:
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

class_name = input().strip()

if class_name == "Dog":
    animal = Dog()
    print(animal.speak())
elif class_name == "Cat":
    animal = Cat()
    print(animal.speak())
```

bài 15:Yêu cầu:
Xây dựng lớp Employee(name, salary).
Xây dựng lớp Manager(name, salary, bonus) kế thừa từ Employee.
Cả hai lớp đều có phương thức income():
- Employee: trả về salary
- Manager: trả về salary + bonus
Input:
Dòng 1: loại đối tượng (Employee hoặc Manager)
Dòng 2: name salary [bonus]
Output:
In ra tổng thu nhập của đối tượng. 

```python
class Employee:
    def __init__(self, name: str, salary: float):
        self.name = name
        self.salary = salary

    def income(self) -> float:
        return self.salary

class Manager(Employee):
    def __init__(self, name: str, salary: float, bonus: float):
        super().__init__(name, salary)
        self.bonus = bonus

    def income(self) -> float:
        return self.salary + self.bonus

obj_type = input().strip()
info = input().split()

name = info[0]
salary = float(info[1])

if obj_type == "Employee":
    emp = Employee(name, salary)
    print(int(emp.income()))
    
elif obj_type == "Manager":
    bonus = float(info[2])
    mgr = Manager(name, salary, bonus)
    print(int(mgr.income()))
```
