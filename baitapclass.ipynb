bài 1:
[Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diệntích toàn phần) và thể tích của hình trụ
Đầu vào
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách.
Đầu ra
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử dụng lại những hàm vừa viết, π=3.14).
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy)

```python
class Cylinder:
    def __init__(self, rr, hh):
        self.r = rr
        self.h = hh
    def dien_tich(self):
        return (2 * (self.r ** 2) * 3.14) + (2 * 3.14 * self.r * self.h)
    def the_tich(self):
        return (3.14 * (self.r ** 2) * self.h)
    
dauvao = list(map(float, input().split()))
rr = dauvao[0]
hh = dauvao[1]
cylinder1 = Cylinder(rr, hh)
rs1 = cylinder1.dien_tich()
rs2 = cylinder1.the_tich()
print(f"{rs1:.2f} {rs2:.2f}")
```

bài 2:[Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương
thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không
và phương thức tính ngày kế tiếp của ngày tháng năm nào đó
nếu nó là hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn
dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in
ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.
Đầu vào
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng
dd/mm/yyyy
.
Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class Date:
  def __init__(self, dd, mm ,yy):
    self.d = dd
    self.m = mm
    self.y = yy
  def check_valid(self):
    max_day = -1

    match self.m:
      case 1| 3| 5| 7| 8| 10| 12:
        max_day = 31
      case 4| 6| 9| 11:
        max_day = 30
      case 2:
        if (self.y % 4 == 0 and self.y % 100 != 0) and self.y % 400 == 0:
          max_day = 29
        else:
          max_day = 28
    return (self.y > 0) and (1 <= self.d <= max_day)

  def get_nextday(self):
    if self.check_valid():
        nextday = Date(self.d + 1, self.m, self.y)
    if nextday.check_valid():
        return nextday
    nextday = Date(1, self.m + 1, self.y)
    if nextday.check_valid():
        return nextday
    return Date(1, 1, self.y+1)

  def __str__(self):
    return f"{self.d:02d}/{self.m:02d}/{self.y}"

day1 = Date(29, 1, 2000)
day2 = Date(29, 2, 2001)
day3 = Date(21, 4, 2001)
print(day1.check_valid())
print(day2.check_valid())
print(day3.check_valid())


print(day3.get_nextday)
      
```

bài 3:[Point]
Xây dựng lớp điểm với hai thuộc tính x, y cùng các phương thức kiểm tra xem có phải gốc tọa dộ không, có nằm trên trục tung không, có nằm trên trục hoành không, tính khoảng cách từ điểm đó tới một điểm khác, 
Đầu vào:
Nhập vào tọa độ một điểm M bất kỳ
Đầu ra:
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ là bao nhiêu (lấy 2 chữ số phần thập phân) 


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def goc_toa_do(self):
        return self.x == 0 and self.y == 0
    def truc_tung(self):
        return self.x == 0 and self.y != 0
    def truc_hoanh(self):
        return self.x != 0 and self.y == 0
    def kc(self, diem_khac):
        return ((self.x - diem_khac.x)**2 + (self.y - diem_khac.y)**2)**0.5
    
M_x = float(input())
M_y = float(input())
M = Point(M_x, M_y)
N_x = float(input("Nhập x cho điểm N: "))
N_y = float(input("Nhập y cho điểm N: "))
N = Point(N_x, N_y)
goc_toa_do = Point(0, 0)

if M.goc_toa_do():
    print(f"M là gốc toạ đọ")
elif M.truc_hoanh():
    print(f"M là trục hoành")
elif M.truc_tung():
    print(f"M là trục tung")
print(f"{M.kc(N):.2f}")
```

bài 4:[Calculator]
Xây dựng lớp Calculator (máy tính bỏ túi) với các phương thức để thực hiện các phép toán 1. add (cộng), 2.subtract(trừ), 3. product (nhân) và 4. divide(chia), 5. power(hàm mũ), 6. mod (lấy phần dư) và phương thức 7. set_numbers(cập nhật 2 số mới) nếu người dùng muốn tính toán trên 2 số khác.
Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính, Mỗi phương thức thực hiện một phép tính và trả về kết quả tương ứng
Tạo menu và cho người dùng tính toán tới khi gõ "yes", "YES" hoặc "Yes" cho câu hỏi "Bạn có muốn thoát?" mới dừng.

```python
class calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2
    
    def set_numbers(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2
    
    def subtract(self):
        return self.num1 - self.num2
    
    def product(self):
        return self.num1 * self.num2
    
    def divide(self):
        if self.num2 != 0:
            return self.num1 / self.num2
        else:
            return "lỗi"
    
    def power(self):
        return self.num1 ** self.num2
    
    def mod(self):
        if self.num2 != 0:
            return self.num1 % self.num2
        else:
            return "lỗi"
        
n1 = float(input())
n2 = float(input())
calc = calculator(n1, n2)

while True:
    print("\n--- MENU MÁY TÍNH ---")
    print("1. Cộng (add)")
    print("2. Trừ (subtract)")
    print("3. Nhân (product)")
    print("4. Chia (divide)")
    print("5. Hàm mũ (power)")
    print("6. Lấy phần dư (mod)")
    print("7. Cập nhật số mới (set_numbers)")

    choice = input("chọn phép tính (1 - 7): ")

    if choice == '1':
        print(f"Kết quả: {calc.add()}")
    elif choice == '2':
        print(f"Kết quả: {calc.subtract()}")
    elif choice == '3':
        print(f"Kết quả: {calc.product()}")
    elif choice == '4':
        print(f"Kết quả: {calc.divide()}")
    elif choice == '5':
        print(f"Kết quả: {calc.power()}")
    elif choice == '6':
        print(f"Kết quả: {calc.mod()}")
    elif choice == '7':
        n1 = float(input("Nhập số thứ nhất mới: "))
        n2 = float(input("Nhập số thứ hai mới: "))
        calc.set_numbers(n1, n2)
        print("Đã cập nhật số mới!")
    else:
        print("Lựa chọn không hợp lệ.")

    quit_choice = input("Bạn có muốn thoát? (Gõ 'yes' để thoát): ").strip().lower()
    if quit_choice == "yes":
        print("Tạm biệt!")
        break
```
